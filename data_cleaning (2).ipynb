```python
# Merapikan kolom target 'Classes' ('fire ' / 'not fire ')

df['Classes'] = df['Classes'].astype(str).str.strip()
df['Classes'] = df['Classes'].apply(lambda x: 'fire' if 'fire' in x and 'not' not in x else 'not fire')
```
